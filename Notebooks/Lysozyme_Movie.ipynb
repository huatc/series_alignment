```python
import os
from PIL import Image

# Folder containing BMP images
input_folder = "../Figures/lysozyme_movie/"
output_gif = "output.gif"

# Get all BMP files and sort them
bmp_files = sorted([
    f for f in os.listdir(input_folder)
    if f.lower().endswith(".bmp")
])

# Load images
images = []
for file in bmp_files:
    path = os.path.join(input_folder, file)
    img = Image.open(path)

    # Convert to RGB (important for GIF compatibility)
    img = img.convert("RGB")
    images.append(img)

# Save as GIF
if images:
    images[0].save(
        output_gif,
        save_all=True,
        append_images=images[1:],
        duration=10,   # milliseconds per frame
        loop=0          # 0 = infinite loop
    )

    print(f"GIF saved as {output_gif}")
else:
    print("No BMP files found.")
```
